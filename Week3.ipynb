```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
fp = './week2/images/pic1.png'
img = cv2.imread(fp)
print(img.shape, type(img))
```

```python
cv2.imshow("xxx",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
plt.figure()
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
colors = ('b','g','r')
for i ,color in enumerate(colors): # enumurate ให้ index ของแต่ละ element ที่ visit มาด้วย เก็บใน i
    hist = cv2.calcHist((img),[i],None,[256],[0,256])
    plt.plot(hist, color = color)
plt.title('Histogram for gray scale image')
plt.show()
```

```python
b,g,r = cv2.split(img.astype(np.float32)/255.0)
print(type(b),b.shape,b.min(),b.max())
```

### Method1 long code

```python
gamma = 0.3
be = 255*(b**gamma)
ge = 255*(g**gamma)
re = 255*(r**gamma)
out = cv2.merge((be,ge,re))
out = out.astype(np.uint8)
```

```python
plt.figure()
plt.imshow(cv2.cvtColor(out,cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
color = ('b','g','r')
for i ,x in enumerate(color): # enumurate ให้ index ของแต่ละ element ที่ visit มาด้วย เก็บใน i
    hist = cv2.calcHist((out),[i],None,[256],[0,256])
    plt.plot(hist, color=color[i])
plt.title('Histogram for gray scale image')
plt.show()
```
