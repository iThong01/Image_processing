```python
import matplotlib.pyplot as plt
import numpy as np
import cv2
```

```python
fp = './pic2.png'
img = cv2.imread(fp,cv2.IMREAD_GRAYSCALE)
img.shape
```

```python
def hist_img(img):
    return np.bincount(img.ravel(), minlength=256)

def plot_hist(hist, T=None, start=0, end=256, title="Histogram"):
    levels = np.arange(256)

    plt.figure(figsize=(7, 3))
    plt.plot(levels[start:end], hist[start:end], color="black")

    if T is not None:
        plt.axvline(T, color="red", linestyle="--", label=f"T = {T}")
        plt.legend()

    plt.title(title)
    plt.xlabel("Pixel intensity")
    plt.ylabel("Number of pixels")
    plt.grid(alpha=0.2)
    plt.show()
```

```python
plt.figure(figsize=(4,4))
plt.title('original')
plt.imshow(img,cmap='gray')
plt.axis('off')
plt.show()

hist_ori = hist_img(img)
plot_hist(hist_ori, title="Hist Original")
```

```python
print(img.min())
print(img.max())
```

```python
min , max = 120 , 200

img_stretched = (img.astype(np.float32)-min) / (max-min) * 255
img_stretched = np.clip(img_stretched,0,255).astype(np.uint8)
```

```python
plt.figure(figsize=(4,4))
plt.title('img_stretched')
plt.imshow(img_stretched,cmap='gray')
plt.axis('off')
plt.show()

hist_ori = hist_img(img_stretched)
plot_hist(hist_ori, title="Hist Original")
```
