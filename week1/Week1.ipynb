```python
import cv2, numpy as np
```

```python
print(cv2.IMREAD_COLOR)
```

```python
im = cv2.imread('./images/images/lena.png',cv2.IMREAD_COLOR)
```

```python
print(type(im),im.shape)
```

```python
cv2.imshow('test',im)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
b, g, r = cv2.split(im)
print(type(b), b.shape)
```

### Code in below is the same split

```python
b = im[:,:,0].astype(np.float16) # astype use to change type
g = im[:,:,1].astype(np.float16)
r = im[:,:,2].astype(np.float16)

#change to gray scale
gray = 0.114*b + 0.587*g + 0.299*r
gray = gray.astype(np.uint8)
```

### end

```python
cv2.imshow('gray',gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
